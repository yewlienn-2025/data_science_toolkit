### **Tutorial** **3**

```python
import numpy as np
```

1. Create a numpy array of size 2 x 2

```python
n1=np.array(([2,2],[3,4]))
print(n1)
```

2. Create a random numpy array of size 3 x 4

```python
n2=np.random.random((3,4))
print(n2)
```

3. Create a numpy array of ones and a numpy array of zeroes

```python
n3=np.ones((2,2))
n4=np.zeros((2,2))
print(n3)
print(n4)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
num4=np.random.random((6,3))
print(num4)
```

```python
#Transpose
num4 =np.transpose(num4)
print(num4)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n=1
p=4
m=3
num5=np.random.random((n,p))
num6=np.random.random((p,m))

A=np.dot(num5,num6)

print("Array 1:\n", num5)
print("\nArray 2:\n", num6)
print("\nDot product:\n", A)
```
